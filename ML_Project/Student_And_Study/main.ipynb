```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
import numpy as np

# Load dataset
data = pd.read_csv("./Student.csv")
# print(data.head())

x = data[['Study_Hours_per_Week']]
y = data['Final_Score']

# Train Model
model = LinearRegression()
model.fit(x,y)
predicted_score = model.predict(x)

# valid regression metrics
mae = mean_absolute_error(y, predicted_score)
mse = mean_squared_error(y, predicted_score)
rmse = np.sqrt(mse)
r2 = r2_score(y, predicted_score)

# Show result
print("Mean Absolute Error (MAE):", round(mae, 2))
print("Mean Squared Error (MSE):", round(mse, 2))
print("Root Mean Squared Error (RMSE):", round(rmse, 2))
print("R^2 Score (Model Accuracy):", round(r2, 4))
```

```python

# histogram
plt.figure(figsize=(10,6))
plt.hist(data["Final_Score"],bins=30,color='skyblue',edgecolor='black')
plt.title("Distrubution of Final Exam Scores")
plt.xlabel("Final Exam Score")
plt.ylabel("Number of students")
plt.grid(True)
plt.show()
```

```python
# Scatter Plot
plt.figure(figsize=(10,6))
plt.scatter(x,y,color='blue',label='Actual Scores')
plt.plot(x,predicted_score,color='red',label='Predicted Scores (Regression Line)')
plt.title("Model Prediction VS Actual Score")
plt.xlabel("Study Hour Per Week")
plt.ylabel("Final Output")
plt.grid(True)
plt.legend()
plt.show()

new_hours = 9
predicted_new_score = model.predict([[new_hours]])
print(f"pPedicted Final Score for {new_hours} Hours is {predicted_new_score} Score")
```
